## **Problem 1: Tip Calculator**

**STEP 1: Understand the Problem**

This program helps restaurant staff calculate a tip and the total bill. They enter the bill amount and the tip percentage. The program shows the tip amount and the total amount to charge.

**STEP 2: Identify Inputs and Outputs**

INPUTS:
The user enters the bill amount as a number.
The user enters the tip percentage as a number.

OUTPUTS:
The program shows the tip amount and the total bill, including the tip.
Both amounts are shown as money, rounded to two decimal places.

**STEP 3: Work It By Hand**

1. Convert the tip percentage to a decimal: 15% = 0.15.
2. Calculate the tip: 45.80 x 0.15 = 6.87.
3. Calculate the total bill: 45.80 + 6.87 = 52.67.
4. Tell the staff member: "The tip is 6.87, and the total bill is 52.67."


**Problem 2: Late Fee Calculator**

**STEP 1: Understand the Problem**
This program helps a librarian calculate the fine for an overdue book. The librarian enters how many days the book is late. The program charges 0.50 for each overdue day, up to a maximum fine of 10.00.

**STEP 2: Identify Inputs and Outputs**

Inputs:
The number of days the book is overdue.

Outputs:

The program calculates the late fee at 0.50 per overdue day.

If the calculated fine is 10.00 or less, the program shows the fine amount.

If the calculated fine is more than 10.00, the program shows "Maximum fine reached."


**STEP 3: Work It By Hand**

**Case A: Book is 5 days late**

My calculation: 5 × 0.50 = 2.50

Fine: 2.50

**Case B: Book is 25 days late**

My calculation: 25 × 0.50 = 12.50

Fine: Maximum fine reached.

Special note: The calculated fine is more than the maximum fine of 10.00, so the maximum-fine message is shown.

**Case C: Book is 0 days late**

My calculation: 0 × 0.50 = 0.00

Fine: 0.00
