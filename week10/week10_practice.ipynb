## Practice Exercise 2.1: Spending Trend Analysis

::: {.callout-tip}
## Business Context

You're helping a friend understand their spending habits over time.

**Challenge**: Ask AI to help you:
1. Convert dates to datetime format
2. Analyze spending trends by day of week
3. Identify if spending has increased over time
4. Spot any unusual spending patterns

**Focus**: Practice asking AI to explain any complex date handling code it provides.
:::

**Your Analysis:**

````python
# Your AI collaboration for time-based analysis
````

````python
import pandas as pd

# Display the transaction data column names
print("Columns in the transaction data:")
print(transactions_df.columns.tolist())

# Try to identify the date and amount columns
date_candidates = [
    "Date", "Transaction_Date", "TransactionDate",
    "date", "transaction_date"
]
amount_candidates = [
    "Amount", "Transaction_Amount", "TransactionAmount",
    "amount", "transaction_amount"
]

date_col = next(
    (col for col in date_candidates if col in transactions_df.columns),
    None
)
amount_col = next(
    (col for col in amount_candidates if col in transactions_df.columns),
    None
)

# If common names were not found, search column names for keywords
if date_col is None:
    date_col = next(
        (col for col in transactions_df.columns if "date" in col.lower()),
        None
    )

if amount_col is None:
    amount_col = next(
        (
            col for col in transactions_df.columns
            if any(word in col.lower() for word in ["amount", "spend", "value"])
        ),
        None
    )

# Check whether both columns were identified
if date_col is None or amount_col is None:
    print("\nCould not automatically identify the date and amount columns.")
    print("Check the column names displayed above.")
else:
    print(f"\nDate column: {date_col}")
    print(f"Amount column: {amount_col}")

    # Convert the date and amount columns to suitable formats
    transactions_df[date_col] = pd.to_datetime(
        transactions_df[date_col],
        errors="coerce"
    )

    transactions_df[amount_col] = pd.to_numeric(
        transactions_df[amount_col]
        .astype(str)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False),
        errors="coerce"
    )

    # Remove rows with missing or invalid dates or amounts
    analysis_df = transactions_df.dropna(
        subset=[date_col, amount_col]
    ).copy()

    # Create day-of-week and month columns
    analysis_df["Day_of_Week"] = analysis_df[date_col].dt.day_name()
    analysis_df["Month"] = analysis_df[date_col].dt.to_period("M").astype(str)

    # Summarise spending by day of the week
    weekday_summary = (
        analysis_df.groupby("Day_of_Week")[amount_col]
        .agg(["sum", "mean", "count"])
        .rename(columns={
            "sum": "Total_Spending",
            "mean": "Average_Spending",
            "count": "Transactions"
        })
        .reindex([
            "Monday", "Tuesday", "Wednesday", "Thursday",
            "Friday", "Saturday", "Sunday"
        ])
        .round(2)
    )

    # Summarise spending by month to observe the trend over time
    monthly_summary = (
        analysis_df.groupby("Month")[amount_col]
        .agg(["sum", "mean", "count"])
        .rename(columns={
            "sum": "Total_Spending",
            "mean": "Average_Spending",
            "count": "Transactions"
        })
        .round(2)
    )

    # Identify unusually high transactions:
    # amounts greater than the mean plus two standard deviations
    average_amount = analysis_df[amount_col].mean()
    standard_deviation = analysis_df[amount_col].std()

    unusual = analysis_df[
        analysis_df[amount_col] > average_amount + 2 * standard_deviation
    ].sort_values(amount_col, ascending=False)

    print("\nSpending by day of the week:")
    display(weekday_summary)

    print("\nSpending by month:")
    display(monthly_summary)

    print("\nUnusually high transactions:")
    if unusual.empty:
        print("No unusually high transactions were found using this rule.")
    else:
        display(unusual[[date_col, amount_col, "Day_of_Week"]])

    # Print a simple summary of the results
    print("\nSimple summary:")

    if not monthly_summary.empty:
        first_month = monthly_summary.iloc[0]["Total_Spending"]
        last_month = monthly_summary.iloc[-1]["Total_Spending"]

        if last_month > first_month:
            print(
                "Total spending was higher in the last month "
                "than in the first month."
            )
        elif last_month < first_month:
            print(
                "Total spending was lower in the last month "
                "than in the first month."
            )
        else:
            print(
                "Total spending was the same in the first and last months."
            )

    if not weekday_summary["Total_Spending"].dropna().empty:
        highest_day = weekday_summary["Total_Spending"].idxmax()
        print(f"The highest total spending was on {highest_day}.")

    print(
        "Limitation: This analysis only covers the transactions in this "
        "dataset. It may not represent spending habits outside the "
        "recorded period."
    )
````

The analysis shows that total spending varied across the days of the week. Tuesday had a negative total spending amount of -10.70. This may be due to a refund or a reversed transaction recorded as a negative value, so it should not be treated as normal spending without checking the transaction details. The dataset only covers August 2024, so it is not enough to determine whether spending increased over time or to identify longer-term patterns.

## Practice Exercise 2.2: Budget Category Analysis

::: {.callout-tip}
## Personal Finance Scenario

Someone wants to set a realistic monthly budget based on their spending history.

**Challenge**: Use AI to help:
1. Calculate average monthly spending by category
2. Identify categories with high variation (inconsistent spending)
3. Suggest realistic budget amounts for each category
4. Find categories where spending might be reduced

**Advanced**: Ask AI to explain the statistical concepts it uses.
:::

**Your Budget Analysis:**

````python
# AI-assisted budget recommendation code

import pandas as pd

# Use the existing transaction data
budget_df = transactions_df.copy()

# Convert the amount column to numeric values
budget_df["Amount"] = pd.to_numeric(
    budget_df["Amount"]
        .astype(str)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False),
    errors="coerce"
)

# Remove rows with missing category or amount
budget_df = budget_df.dropna(subset=["Category", "Amount"]).copy()

# Summarize spending by category
category_summary = (
    budget_df.groupby("Category")["Amount"]
    .agg(
        Total_Spending="sum",
        Average_Transaction="mean",
        Transactions="count",
        Spending_Std="std"
    )
    .fillna({"Spending_Std": 0})
    .sort_values("Total_Spending", ascending=False)
)

print("Spending by category:")
display(category_summary)

# Calculate category shares using positive spending only
positive_spending = budget_df.loc[budget_df["Amount"] > 0].copy()
total_positive_spending = positive_spending["Amount"].sum()

if total_positive_spending > 0:
    category_share = (
        positive_spending.groupby("Category")["Amount"].sum()
        / total_positive_spending * 100
    ).sort_values(ascending=False)

    print("\nShare of positive spending by category (%):")
    display(category_share.round(1))

# Identify categories with the highest and lowest total amounts
if not category_summary.empty:
    highest_category = category_summary["Total_Spending"].idxmax()
    lowest_category = category_summary["Total_Spending"].idxmin()

    print(f"\nHighest total category: {highest_category}")
    print(f"Lowest total category: {lowest_category}")

print(
    "\nLimitation: The dataset covers only one month, "
    "so it cannot show reliable monthly averages or long-term variation."
)
````

Budget Category Analysis: Groceries had the highest spending, with a total of 394 dollars and 70 cents. This was 43.3 percent of positive spending. Entertainment was second, at 206 dollars and 95 cents, or 22.7 percent. Dining was third, at 113 dollars and 90 cents, or 12.5 percent. Groceries and entertainment had the largest variation in transaction amounts. Their standard deviations were about 49.05 and 40.77, respectively. This suggests that spending in these categories was less consistent during the recorded month.

Based on this one-month record, a starting monthly budget could be approximately:

Groceries: 435 dollars
Entertainment: 230 dollars
Dining: 125 dollars
Transport: 100 dollars
Utilities: 75 dollars
Coffee: 42 dollars
These amounts are slightly above the recorded spending totals to allow some flexibility. To reduce spending, the person could first review entertainment and dining purchases, while keeping track of grocery spending because it is the largest category.

Refunds totalled negative 37.50 dollars and should be treated separately from ordinary spending. This dataset covers only August 2024, so these budget amounts are initial estimates, not reliable long-term averages. More months of data are needed to confirm usual spending and variation.

**Your Business Analysis:**

````python
# Practice comprehensive business analysis with AI assistance


import pandas as pd

# Load the sales data
sales_df = pd.read_csv("/content/sales_data.csv")

# Convert currency-formatted values into numbers
for column in ["Quantity", "Unit_Price", "Total_Sale"]:
    sales_df[column] = pd.to_numeric(
        sales_df[column]
        .astype(str)
        .str.replace(r"[$,]", "", regex=True),
        errors="coerce"
    )

# Remove rows with missing values needed for this analysis
analysis_df = sales_df.dropna(
    subset=["Region", "Product", "Quantity", "Total_Sale"]
).copy()

# Treat each region as a store location
store_summary = (
    analysis_df.groupby("Region")
    .agg(
        Total_Sales=("Total_Sale", "sum"),
        Average_Sale=("Total_Sale", "mean"),
        Transaction_Count=("Total_Sale", "count")
    )
    .sort_values("Total_Sales", ascending=False)
    .round(2)
)

# Summarise sales and quantities by product
product_summary = (
    analysis_df.groupby("Product")
    .agg(
        Total_Sales=("Total_Sale", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Transaction_Count=("Product", "count")
    )
    .sort_values("Total_Sales", ascending=False)
    .round(2)
)

print("Store performance (regions treated as store locations):")
display(store_summary)

print("\nProduct performance:")
display(product_summary)

# Focus on actionable insights, not just data summaries
````

````python
# AI-assisted personal finance advice generationimport os
import re
import sys
import pandas as pd

# Load the transactions file if it is not already loaded.
if "transactions_df" not in globals() or not isinstance(
    globals().get("transactions_df"), pd.DataFrame
):
    file_path = "/content/transactions.csv"
    if not os.path.exists(file_path):
        print(f"Could not find {file_path}. Upload transactions.csv to Colab and try again.")
        raise FileNotFoundError(file_path)
    transactions_df = pd.read_csv(file_path)

# Match likely column names, allowing common spelling and formatting variations.
def normalize_name(name):
    return re.sub(r"[^a-z0-9]", "", str(name).lower())

columns_by_normalized_name = {
    normalize_name(column): column for column in transactions_df.columns
}

def find_column(candidates, keywords):
    for candidate in candidates:
        match = columns_by_normalized_name.get(normalize_name(candidate))
        if match is not None:
            return match

    for column in transactions_df.columns:
        normalized = normalize_name(column)
        if any(keyword in normalized for keyword in keywords):
            return column
    return None

category_col = find_column(
    ["category", "spending_category", "expense_category", "type", "description"],
    ["category", "expense", "merchant", "description", "type"],
)
amount_col = find_column(
    ["amount", "transaction_amount", "debit", "spend", "value", "cost"],
    ["amount", "debit", "spend", "cost", "value"],
)
date_col = find_column(
    ["date", "transaction_date", "posted_date", "datetime", "timestamp"],
    ["date", "time"],
)

missing_columns = []
if category_col is None:
    missing_columns.append("category")
if amount_col is None:
    missing_columns.append("amount")
if date_col is None:
    missing_columns.append("date")

if missing_columns:
    print("I could not identify the required column(s): " + ", ".join(missing_columns))
    print("Available columns:", list(transactions_df.columns))
    raise SystemExit("Please check your column names or update the column matching rules.")

# Convert amounts such as "$1,234.56" and "($20.00)" to numbers.
amount_text = transactions_df[amount_col].astype(str).str.strip()
parentheses_are_negative = amount_text.str.match(r"^\(.*\)$")
clean_amount = (
    amount_text
    .str.replace(r"[$£€¥,\s]", "", regex=True)
    .str.replace(r"^\((.*)\)$", r"-\1", regex=True)
    .str.replace(r"[^0-9.+\-]", "", regex=True)
)
transactions_df["_clean_amount"] = pd.to_numeric(clean_amount, errors="coerce")
transactions_df.loc[parentheses_are_negative, "_clean_amount"] = (
    -transactions_df.loc[parentheses_are_negative, "_clean_amount"].abs()
)

# Parse dates and report the period represented by usable records.
transactions_df["_clean_date"] = pd.to_datetime(
    transactions_df[date_col], errors="coerce"
)
valid_dates = transactions_df["_clean_date"].dropna()

print("PERSONAL FINANCE SPENDING CHECK")
print("-" * 38)
if not valid_dates.empty:
    print(
        "Data period:",
        f"{valid_dates.min().date()} to {valid_dates.max().date()}",
    )
else:
    print("Data period: dates could not be parsed.")

# Keep categorized positive amounts for spending totals.
analysis_df = transactions_df[
    transactions_df["_clean_amount"].notna()
    & transactions_df[category_col].notna()
    & (transactions_df["_clean_amount"] > 0)
].copy()

if analysis_df.empty:
    print("\nNo positive spending amounts with categories were found.")
    print("Check that purchases are represented as positive amounts in the file.")
    raise SystemExit

analysis_df["_category"] = analysis_df[category_col].astype(str).str.strip()
analysis_df = analysis_df[analysis_df["_category"] != ""]

category_spending = analysis_df.groupby("_category")["_clean_amount"].sum()
total_spending = category_spending.sum()

if total_spending <= 0:
    print("\nThere is no positive spending to summarize.")
    raise SystemExit

category_percentages = category_spending / total_spending * 100

# Illustrative examples only; adjust these percentages to fit your circumstances.
# The example shares add up to 100%.
example_benchmarks = {
    "housing": 30,
    "rent": 30,
    "food": 15,
    "groceries": 15,
    "transportation": 10,
    "transport": 10,
    "utilities": 8,
    "health": 7,
    "healthcare": 7,
    "entertainment": 10,
    "dining": 5,
    "shopping": 5,
    "savings": 10,
}

def benchmark_for(category):
    key = normalize_name(category)
    for benchmark_name, percentage in example_benchmarks.items():
        if key == normalize_name(benchmark_name):
            return percentage
    return None

print(f"\nPositive spending found: {total_spending:,.2f}")
print("\nCategory comparison")
print("(Benchmarks are illustrative examples, not personal recommendations.)")
print("-" * 72)
print(f"{'Category':<24} {'Spent':>12} {'Share':>10} {'Example':>10}  Note")
print("-" * 72)

for category, spent in category_spending.sort_values(ascending=False).items():
    share = category_percentages[category]
    benchmark = benchmark_for(category)

    if benchmark is None:
        note = "No matching example benchmark"
        benchmark_text = "—"
    elif share > benchmark:
        note = "Above example"
        benchmark_text = f"{benchmark:.0f}%"
    else:
        note = "Within/below example"
        benchmark_text = f"{benchmark:.0f}%"

    print(
        f"{category[:23]:<24} {spent:>12,.2f} "
        f"{share:>9.1f}% {benchmark_text:>10}  {note}"
    )

print("\nPotential next steps")
above_categories = [
    (category, category_percentages[category], benchmark_for(category))
    for category in category_spending.index
    if benchmark_for(category) is not None
    and category_percentages[category] > benchmark_for(category)
]

if above_categories:
    for category, share, benchmark in sorted(
        above_categories, key=lambda item: item[1] - item[2], reverse=True
    ):
        category_key = normalize_name(category)
        if any(word in category_key for word in ["dining", "restaurant", "food"]):
            action = "Try planning a few lower-cost meals or setting a weekly dining limit."
        elif any(word in category_key for word in ["shopping", "retail"]):
            action = "Consider a 24-hour pause before non-essential purchases."
        elif any(word in category_key for word in ["transport", "travel", "fuel"]):
            action = "Compare transit, carpool, or trip-planning options where practical."
        elif any(word in category_key for word in ["entertainment", "subscription"]):
            action = "Review recurring subscriptions and keep the ones you use most."
        elif any(word in category_key for word in ["housing", "rent"]):
            action = "Review recurring housing costs and possible lower-cost options when feasible."
        else:
            action = "Review recent transactions and look for one manageable cost to reduce."

        print(
            f"- {category}: {share:.1f}% of spending, above the "
            f"{benchmark:.0f}% example. {action}"
        )
else:
    print("No categories matched an example benchmark above its illustrative percentage.")

print("\nEncouragement")
within_or_below = [
    category
    for category in category_spending.index
    if benchmark_for(category) is not None
    and category_percentages[category] <= benchmark_for(category)
]
if within_or_below:
    print(
        "Nice work: these matched categories are within or below their examples: "
        + ", ".join(within_or_below)
        + ". Keep focusing on choices that fit your needs."
    )
else:
    print(
        "Keep going: a category comparison is only a starting point. "
        "Small, realistic changes can help you understand your spending."
    )

print(
    "\nImportant: Results reflect only the transactions and dates in this file. "
    "The percentages are limited illustrative examples, not a universal budget, "
    "and categories without a matching example cannot be compared. "
    "This output is for learning and is not financial advice."
)
````
