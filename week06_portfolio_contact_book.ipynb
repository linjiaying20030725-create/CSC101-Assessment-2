# Week 7 Portfolio - Mini Contact Book

This notebook creates a simple contact book using a Python dictionary.
I worked with AI to build and test this code.


```python
# Step 1: Create a dictionary with 3 contacts
# The name is the key, and the phone number is the value
contacts = {
    "Alice": "0412 345 678",
    "Bob": "0400 111 222",
    "Charlie": "0433 999 888"
}

print(contacts)
```

```python
# Step 2: Add a new contact
contacts["Jiaying"] = "0455 123 456"

print(contacts)
```

```python
# Step 3: Update Bob's phone number
contacts["Bob"] = "0488 777 666"

print(contacts)
```

```python
# Step 4: Loop through the dictionary and print each contact
for name, phone in contacts.items():
    print(name + ", " + phone)
```

```python
# Part B: Nested dictionary
# Each contact stores both a phone number and an email address
contacts_v2 = {
    "Alice":   {"phone": "0412 345 678", "email": "alice@example.com"},
    "Bob":     {"phone": "0488 777 666", "email": "bob@example.com"},
    "Charlie": {"phone": "0433 999 888", "email": "charlie@example.com"}
}

for name, details in contacts_v2.items():
    print(name + ", " + details["phone"] + ", " + details["email"])
```

```python
# Optional: ask for a name and print that person's details
name = input("Enter a name: ")

if name in contacts_v2:
    print(name + ", " + contacts_v2[name]["phone"] + ", " + contacts_v2[name]["email"])
else:
    print("Sorry, this name is not in the contact book.")
```
