# Week 9: Smart Personal Finance Manager

This project reads transaction data from a text file, processes valid transactions, skips corrupted data, and displays a financial summary.

The program uses separate functions so that each part of the task is easier to understand and maintain.


```python
# Open the transaction file and read all lines
with open("transactions.txt", "r") as file:
    lines = file.readlines()

print(lines)
```

```python
def load_transactions(filename):
    """Read lines from a transaction file safely."""
    try:
        with open(filename, "r") as file:
            lines = file.readlines()

        return [line.strip() for line in lines if line.strip()]

    except FileNotFoundError:
        print(f"File '{filename}' not found!")
        return []
```

```python
# Load and display the transaction lines from the file
raw_lines = load_transactions("transactions.txt")

print("Transaction lines loaded:")
print(raw_lines)
print(f"Number of lines loaded: {len(raw_lines)}")
```

```python
def process_transactions(lines):
    """Convert transaction lines into dictionaries and skip invalid lines."""

    transactions = []
    failed_count = 0

    # Process each line in the input file
    for line in lines:
        try:
            # Skip blank lines
            if not line.strip():
                failed_count += 1
                print("Skipping empty line")
                continue

            # Split each line into a description and an amount
            parts = line.split(",")

            if len(parts) != 2:
                raise ValueError("The line must contain a description and an amount.")

            # Clean the text and convert the amount to a number
            description = parts[0].strip()
            amount = float(parts[1].strip())

            if not description:
                raise ValueError("The description is empty.")

            # Store the valid transaction as a dictionary
            transaction = {
                "description": description,
                "amount": amount
            }

            transactions.append(transaction)

        except (IndexError, ValueError):
            # Count and report invalid transaction lines
            failed_count += 1
            print(f"Skipping invalid line: {line}")

    print(f"Number of invalid lines skipped: {failed_count}")
    return transactions
```

```python
# Test the transaction-processing function
transactions = process_transactions(raw_lines)

print("\nValid transactions:")

# Display each valid transaction
for transaction in transactions:
    print(transaction)

print(f"\nNumber of valid transactions: {len(transactions)}")
```

```python
def summarize_transactions(transactions):
    """
    Calculate income, expenses, and net amount from valid transactions.
    """

    total_income = 0
    total_expenses = 0

    # Process each valid transaction
    for transaction in transactions:
        amount = transaction["amount"]

        # Classify positive amounts as income
        if amount >= 0:
            total_income += amount
        else:
            # Convert negative expenses to positive values
            total_expenses += abs(amount)

    # Calculate the final amount after expenses
    net_amount = total_income - total_expenses

    summary = {
        "total_income": total_income,
        "total_expenses": total_expenses,
        "net_amount": net_amount
    }

    return summary
```

```python
summary = summarize_transactions(transactions)

print("Transaction Summary")
print("-------------------")
print(f"Total income: ${summary['total_income']:.2f}")
print(f"Total expenses: ${summary['total_expenses']:.2f}")
print(f"Net amount: ${summary['net_amount']:.2f}")
```

```python
raw_lines = load_transactions("transactions.txt")
transactions = process_transactions(raw_lines)
summary = summarize_transactions(transactions)

print("Final Report")
print("------------")
print(f"Valid transactions: {len(transactions)}")
print(f"Invalid lines skipped: {len(raw_lines) - len(transactions)}")
print(f"Total income: ${summary['total_income']:.2f}")
print(f"Total expenses: ${summary['total_expenses']:.2f}")
print(f"Net amount: ${summary['net_amount']:.2f}")
```
