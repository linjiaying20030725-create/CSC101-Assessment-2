# Week 9: Smart Personal Finance Manager

This project reads transaction data from a text file, processes valid transactions, skips corrupted data, and displays a financial summary.

The program uses separate functions so that each part of the task is easier to understand and maintain.


```python
# 打开 transactions.txt 文件并读取所有内容
with open("transactions.txt", "r") as file:
    lines = file.readlines()

# 显示文件内容
print(lines)
```

```python
def load_transactions(filename):
    """Read lines from a transaction file safely."""
    try:
        with open(filename, "r") as file:
            lines = file.readlines()
        return [line.strip() for line in lines if line.strip()]
    except FileNotFoundError:
        print(f"File '{filename}' not found!")
        return []
```

```python
raw_lines = load_transactions("transactions.txt")

print("Transaction lines loaded:")
print(raw_lines)
print(f"Number of lines loaded: {len(raw_lines)}")
```

```python
def process_transactions(lines):
    """Convert valid transaction lines into dictionaries."""

    transactions = []
    failed_count = 0

    for line in lines:
        try:
            if not line.strip():
                failed_count += 1
                print("Skipping empty line")
                continue

            parts = line.split(",")

            if len(parts) != 2:
                raise ValueError("The line must contain a description and an amount.")

            description = parts[0].strip()
            amount = float(parts[1].strip())

            if not description:
                raise ValueError("The description is empty.")

            transaction = {
                "description": description,
                "amount": amount
            }

            transactions.append(transaction)

        except (IndexError, ValueError):
            failed_count += 1
            print(f"Skipping invalid line: {line}")

    print(f"Number of invalid lines skipped: {failed_count}")
    return transactions
```

```python
transactions = process_transactions(raw_lines)

print("\nValid transactions:")
for transaction in transactions:
    print(transaction)

print(f"\nNumber of valid transactions: {len(transactions)}")
```
